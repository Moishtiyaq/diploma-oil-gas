```python
import math

def average_angle_tvd(md1, inc1_deg, md2, inc2_deg, tvd1):
    """
    Calculate TVD at current depth using Average Angle Method.

    Parameters:
    md1 (float): Previous measured depth
    inc1_deg (float): Previous inclination in degrees
    md2 (float): Current measured depth
    inc2_deg (float): Current inclination in degrees
    tvd1 (float): TVD at previous station

    Returns:
    float: TVD at current station
    """
    # Convert inclination to radians
    inc1_rad = math.radians(inc1_deg)
    inc2_rad = math.radians(inc2_deg)

    # Calculate average inclination
    avg_inclination_rad = (inc1_rad + inc2_rad) / 2

    # Delta MD between survey stations
    delta_md = md2 - md1

    # Compute change in TVD
    delta_tvd = delta_md * math.cos(avg_inclination_rad)

    # Final TVD
    tvd2 = tvd1 + delta_tvd
    return tvd2
    

# Example usage
#md1 = 5610      # Previous measured depth
#inc1 = 26.73       # Previous inclination in degrees
#md2 = 5704      # Current measured depth
#inc2 = 28.75       # Current inclination in degrees
#tvd1 = 5092.48     # TVD at previous station

md1 = float(input("Previous measured depth(meter): "))
incl = float(input("Previous inclination in degrees: "))
tvd1 = float(input("TVD at previous station(meter): "))
md2 = float(input("Current measured depth(meter): "))
inc2 = float(input("Current inclination in degrees: "))

tvd2 = average_angle_tvd(md1, inc1, md2, inc2, tvd1)
print(f"Calculated TVD at current station: {tvd2: .2f} meters")
```
